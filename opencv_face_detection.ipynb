```python
import cv2
import matplotlib.pyplot as plt
%matplotlib widget
```

```python
def ishow(image):
    cv2.imshow('image', image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
img = cv2.imread('face.jpg')
```

```python
cam = cv2.VideoCapture(0)
```

```python
_ , img_webcam=cam.read()
ishow(img_webcam)
```

```python
img=img_webcam
```

```python
ishow(img)
```

```python
face_model = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_frontalface_default.xml")
eye_model = cv2.CascadeClassifier(cv2.data.haarcascades + "haarcascade_eye.xml")
```

```python
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
```

```python
ishow(gray)
```

```python
face = face_model.detectMultiScale(
    gray
)
eye = eye_model.detectMultiScale(
    gray
)
```

```python
ishow(out1)
```

```python
x = face[0][0]
y = face[0][1]
x2 = x + face[0][2]
y2 = y + face[0][3]
```

```python
x_eye = eye[1][0]
y_eye = eye[1][1]
x_eye2 = x_eye + eye[1][2]
y_eye2 =y_eye + eye[1][3]

x_eye_2 = eye[0][0]
y_eye_2 = eye[0][1]
x_eye_2_2 = x_eye_2 + eye[0][2]
y_eye_2_2 =y_eye_2 + eye[0][3]
```

```python
img2 = img.copy()
out1 = cv2.rectangle(img2 , (x,y),(x2,y2),(0,250,0),3)
out2 = cv2.rectangle(img2 ,(x_eye,y_eye),(x_eye2,y_eye2),(0,250,0),3 )
out3 = cv2.rectangle(img2 , (x_eye_2,y_eye_2), (x_eye_2_2,y_eye_2_2),(0,250,0),3)
```

```python
ishow(img2)
```

```python
loop=True

while loop:

    _ , img=cam.read()
    gray = cv2.cvtColor(img,cv2.COLOR_BGR2GRAY)
    face=face_model.detectMultiScale(gray)
    #gray_face=gray[]
    eye=eye_model.detectMultiScale(gray)
    if len(face)==0:
        cv2.imshow('Webcam', img)
        if cv2.waitKey(1) == ord('q'):
            cv2.destroyAllWindows()
            break
        continue
    img3 = img.copy()
    x=face[0][0]
    y=face[0][1]
    x2=x + face[0][2]
    y2=y + face[0][3]

    out=cv2.rectangle(img3,(x,y),(x2,y2),(0,100,255),2)

    for(xe,ye,w,h) in eye:
        cv2.rectangle(img3, (xe,ye), (xe+w, ye+h), (0,255,0),3)

    cv2.imshow('Webcam',out)
    if cv2.waitKey(1) == ord('q'):
        loop=False
        cv2.destroyAllWindows()
        break
```
